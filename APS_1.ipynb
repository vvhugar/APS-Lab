1. A fair coin is tossed 100 times. Calculate:
- Number of heads
- Number of tails
- Experimental probability of Head
- Experimental probability of Tail

```python
import random #needed as we do not know the no of heads and tails we will get during the experiment

trials = 100 #we are tossing 100 times so trial count will be 100
num_heads = 0 #initialise to 0
num_tails = 0

for i in range(trials):
  toss = random.choice(['Heads','Tails']) #the toss can be either heads or tails
  if toss=='Heads':
    num_heads += 1 #increment for heads or tails
  else: #then if toss is tails
    num_tails += 1
```

```python
print("Number of heads: ",num_heads)
print("Number of tails: ",num_tails)
print("Experimental probability of Head: ",num_heads/trials) #p(h)=num(h)/num(total)
print("Experimental probability of Tail: ",num_tails/trials)
```

3. A fair die is rolled 6000 times. Calculate the experimental probability of each outcome.

```python
import random

rolls = 6000 # we are rolling 6000 times
one = 0 # initially all will be 0
two = 0
three = 0
four = 0
five = 0
six = 0

for i in range(rolls): #for the range of trails increment for each possible outcome
  roll = random.randint(1,6) #roll can be between 1 to 6
  if roll==1:
    one += 1
  elif roll==2:
    two += 1
  elif roll==3:
    three += 1
  elif roll==4:
    four += 1
  elif roll==5:
    five += 1
  else:
    six += 1

print('Experimental probability of rolling 1 is', one/rolls)
print('Experimental probability of rolling 2 is', two/rolls)
print('Experimental probability of rolling 3 is', three/rolls)
print('Experimental probability of rolling 4 is', four/rolls)
print('Experimental probability of rolling 5 is', five/rolls)
print('Experimental probability of rolling 6 is', six/rolls)
```

4. **Addition rule for disjoint events.** A die is rolled 10000 times. Define:
- A = getting 1
- B = getting 6
- Check the events are joint or disjoint
- Verify (AuB)=P(A)+P(B)

```python
import random

rolls = 10000
one = 0
six = 0

for i in range(rolls):
  roll = random.randint(1,6)
  if roll==1:
    one += 1
  elif roll==6:
    six += 1

print('A = probability of gtting 1 is', one/rolls)
print('B = probability of getting 6 is', six/rolls)
```

```python
#check the events are joint or disjoint
#disjoint means mutually exclusive aka in the same roll we cannot get 1 as well as 6
A = one/rolls
B = six/rolls

print('P(A) = ',one/rolls)
print('P(B) = ',six/rolls)
print('P(AuB) = ',(one+six)/rolls) # probability of getting either 1 or 6, addition rule
print('P(A) + P(B) = ',A+B)

if A+B==(one+six)/rolls:
  print('Events are disjoint and therefore, P(AuB) = P(A) + P(B) is verified')
else:
  print('Events are not disjoint')
```

5. **Addition rule for overlapping events.** For one die roll, define:
- A = getting an even number, A = {2,4,6}
- B = getting a number greater than 3, B = {4,5,6}
- Verify: P(AuB) = P(A) + P(B) - P (AnB)


```python
import random

rolls = 1000
even = 0
gthan3 = 0
either = 0
both = 0

for i in range(rolls):
  roll = random.randint(1,6) #6 possible outcomes
  iseven = (roll%2==0)
  isgthan3 = (roll>3)

  if iseven:
    even += 1 # if even so{2,4,6}
  if isgthan3:
    gthan3 += 1 # if greater than 3 {4,5,6}
  if iseven or isgthan3:
    either +=1 # if it's even OR greater than 3 so {2,4,5,6}
  if iseven and isgthan3:
    both +=1 # if it's even AND greater than 3 so {4,6}

p_a = even / rolls
p_b = gthan3 / rolls
p_anb = both / rolls
p_aub = either / rolls

print('P(A) = ',p_a)
print('P(B) = ',p_b)
print('P(AuB) = ',p_aub)
print('P(AnB) = ',p_anb)

res = p_a + p_b - p_anb

if p_aub==res:
  print('P(AuB) = P(A) + P(B) - P(AnB)')
  print('Addition rule is verified')
else:
  print('Addition rule is not verified')

#print(f'{p_aub:.4f} = {p_a:.4f} + {p_b:.4f} - {p_anb:.4f}'
```

6. **Multiplication rule for independent events.**
A coin is tossed and a die is rolled 10,000 times. Define:
- Obtaining Head
- Obtaining an even number
- Verify: P(A∩B)=P(A)P(B)

```python
import random

trials = 10000
heads = 0
tails = 0
even_rolls = 0
odd_rolls = 0
both_head_and_even = 0

for i in range(trials):
  # Coin toss
  coin_toss = random.choice(['H', 'T'])
  if coin_toss == 'H':
    heads += 1
  else:
    tails += 1

  # Die roll
  die_roll = random.randint(1, 6)
  if die_roll % 2 == 0: # Check if even
    even_rolls += 1
  else:
    odd_rolls += 1

  # Check for both
  if coin_toss == 'H' and die_roll % 2 == 0:
    both_head_and_even += 1

p_a = heads / trials
p_b = even_rolls / trials
p_anb = both_head_and_even / trials
paXpb = p_a * p_b

print('P(A) (Obtaining Head): ', p_a)
print('P(B) (Obtaining an even number): ', p_b)
print('P(A ∩ B) (Head AND even): ', p_anb)
print('P(A) * P(B): ', paXpb)

# Verify the multiplication rule
# Allow for a small floating point tolerance

if abs(p_anb - paXpb) < 0.01:
  print('The multiplication rule P(A ∩ B) ≈ P(A) * P(B) is verified.')
else:
  print('The multiplication rule is not verified')
```

2. Revised: Taking a total of 10000 outcomes and having a choice between full 10k, 1k, 100 and 10 trails from those.

```python
import random

# Generate 10,000 coin toss outcomes once
full_trials = 10000
toss = [random.choice(['Heads', 'Tails']) for i in range(full_trials)]

sample_sizes = [10, 100, 1000, full_trials] # include the full 10000

for current_sample_size in sample_sizes:
  # selects a subset of outcomes for the current sample size
  # We don't replace, as each toss is a distinct event in the 10000
  outcome = random.sample(toss, current_sample_size) #for toss with any sample size

  num_heads = outcome.count('Heads')
  num_tails = outcome.count('Tails')

  print(f"\n--- Simulation for {current_sample_size} Trials (sampled from {full_trials} total outcomes) ---")
  print("Number of heads: ", num_heads)
  print("Number of tails: ", num_tails)
  print("Experimental probability of Head: ", num_heads / current_sample_size)
  print("Experimental probability of Tail: ", num_tails / current_sample_size)
```

# 1. Birthday Problem
In probability theory, the birthday problem asks for the probability that, in a set of n randomly chosen people, at least two will share the same birthday.

For n=3 find the probability.

```python
n = int(input("Enter the number of people: ")) #user can enter no of people

totaldays = 365

p_no_shared = 1.0 #initially it will be 1 and then we multiply this for each iteration later
for i in range(n):
    p_no_shared *= (totaldays - i) / totaldays #first we should calculate p of not sharing bday aka (365/365) * (364/365) * (363/365) ... and so on for n = 1,2,3 ... and so on

p_2_ppl_share_bday = 1 - p_no_shared # p of sharing bday is 1 - p of not sharing bday

print(f"For {n} people, the probability that at least two share a birthday is: {p_2_ppl_share_bday: }")
```

Let's calculate for up to 30 people and store the values in a list and print a graph

```python
import matplotlib.pyplot as plt
import numpy as np

no_ppl = np.arange(1, 31) # From 1 to 30 people

probabilities_for_graph = [] #first, the list will be empty then values will get added

for num in no_ppl: # from 1 to 30
    totaldays = 365
    p_no_shared = 1.0

    for i in range(num): #for every num
        p_no_shared *= (totaldays - i) / totaldays

    p_2_ppl_share_bday = 1 - p_no_shared
    probabilities_for_graph.append(p_2_ppl_share_bday)

# Plotting the results
plt.plot(no_ppl, probabilities_for_graph) # x and y axis
plt.title('Birthday Problem: Probability of Shared Birthday (n=1 to 30)')
plt.xlabel('Number of People (n)')
plt.ylabel('Probability')
plt.grid(True)
plt.show()
```

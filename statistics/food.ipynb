# Graphical Representation

```python
import seaborn as sns
```

Histogram

```python
df["Order_Cost_INR"].plot(
    kind="hist",
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Food Delivery Order Costs")
plt.xlabel("Order Cost (₹)")
plt.ylabel("Frequency")
plt.show()
```

Box Plot

```python
sns.boxplot(x=df["Order_Cost_INR"])

plt.title("Box Plot of Order Costs")
plt.xlabel("Order Cost (₹)")
plt.show()
```

Scatter Plot

```python
sns.scatterplot(
    data=df,
    x="Delivery_Distance_KM",
    y="Delivery_Time_Minutes"
)

plt.title("Delivery Distance vs Delivery Time")
plt.xlabel("Delivery Distance (km)")
plt.ylabel("Delivery Time (minutes)")
plt.show()
```

Bar Chart

```python
df["Number_of_Items"].value_counts().sort_index().plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Number of Items per Order")
plt.xlabel("Number of Items")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.show()
```

All features

```python
df.hist(figsize=(12, 8), bins=10, edgecolor="black")

plt.suptitle("Distribution of Food Delivery Features")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for column, ax in zip(df.columns, axes.flatten()):
    sns.boxplot(x=df[column], ax=ax)
    ax.set_title(column)

plt.tight_layout()
plt.show()
```

Summary Table

```python
summary = pd.DataFrame({
    "Mean": df.mean(),
    "Median": df.median(),
    "Mode": df.mode().iloc[0],
    "Range": df.max() - df.min(),
    "Variance": df.var(),
    "Standard Deviation": df.std(),
    "IQR": df.quantile(0.75) - df.quantile(0.25),
    "Skewness": df.skew(),
    "Kurtosis": df.kurtosis()
})

print(summary.round(2))
```

Pairplot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df)

plt.suptitle("Pair Plot of Food Delivery Data", y=1.02)
plt.show()
```

Heatmap

```python
# Calculate correlation matrix
correlation = df.corr(numeric_only=True)

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Food Delivery Data")
plt.show()
```

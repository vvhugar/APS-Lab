#Question: Generate 10000 random samples from normal distribution, calculate Population mean, Population standard deviation. Take 30 samples and calculate Sample Mean and Sample standard deviation.

```python
import numpy as np
import pandas as pd
np.random.seed(42)
```

Generate 10,000 random samples from a normal distribution

```python
population = np.random.normal(loc=0, scale=1, size=10000)
```

Calculate Population mean and standard deviation

```python
pop_mean = np.mean(population)
pop_std = np.std(population)

print("--- Population Statistics (N=10000) ---")
print(f"Population Mean: {pop_mean:.6f}")
print(f"Population Standard Deviation: {pop_std:.6f}\n")
```

Take a sample of 30 from the population

```python
sample_size = 30
sample = np.random.choice(population, size=sample_size, replace=False)
```

Calculate Sample Mean and Sample standard deviation

```python
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

print("--- Sample Statistics (n=30) ---")
print(f"Sample Mean: {sample_mean:.6f}")
print(f"Sample Standard Deviation: {sample_std:.6f}")
```
